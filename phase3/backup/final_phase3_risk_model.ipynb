
# Phase 3 — Model Development

This notebook evaluates **four machine learning algorithms**:

1. Logistic Regression (Baseline)
2. Decision Tree
3. Random Forest
4. Gradient Boosting

Workflow:

• Feature selection including **age, gender, department**  
• **Leakage-safe time-based train/test split**  
• **ColumnTransformer** preprocessing for categorical variables  
• **Pipeline integration**  
• Evaluation of all models  
• **Model comparison table**  
• **GridSearchCV hyperparameter tuning**  
• Saving the best model artifact


```python
# ---- Core data manipulation ----
import pandas as pd
import numpy as np

# ---- Model selection & pipeline utilities ----
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# ---- Classifier candidates ----
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import RandomizedSearchCV

# ---- Evaluation metrics ----
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix

# ---- Model persistence ----
import joblib
```

## Load Dataset

```python
# Load the feature-engineered modeling dataset produced in Phase 2 (01_eda.ipynb)
df = pd.read_csv("./data/model_table.csv")
df.head()
```

## Feature Selection

```python
target = "risk_score"   # Multi-class label: Low / Medium / High

# Features defined in risk_features.json — must match exactly at inference time.
# Note: age_group is intentionally omitted; it is derived from age
#       automatically by main.py (_derive_age_group) before the model is called.
features = [
    "age",
    "gender",
    "department",
    "visit_type",
    "chronic_flag",
    "length_of_stay_hours",
    "visit_frequency",
    "avg_los_per_patient",
    "days_since_registration"
    ]
```

## Time-based Train/Test Split

```python
# Sort chronologically — earlier visits become training data, later visits become test.
# A random shuffle would leak future information into training (temporal leakage).
df = df.sort_values("visit_date")

# 80% earliest records → training  |  20% most-recent → test
split_index = int(len(df)*0.8)

train = df.iloc[:split_index]
test  = df.iloc[split_index:]

# fillna(0) handles any residual missing feature values safely
X_train = train[features].fillna(0)
X_test  = test[features].fillna(0)

y_train = train[target]
y_test  = test[target]
```

## ColumnTransformer Preprocessing

```python
# String columns that require encoding before any algorithm can process them
categorical_features = ["gender", "department", "visit_type"]

# Numeric columns — passed through unchanged (tree models don't need scaling)
numerical_features = [
    "age",
    "length_of_stay_hours",
    "visit_frequency",
    "avg_los_per_patient",
    "days_since_registration"
]

# ColumnTransformer applies different transforms to different column groups:
#   "cat" branch: OneHotEncoder converts each category to binary indicator columns.
#                 handle_unknown="ignore" prevents errors if a new category appears at inference.
#   "num" branch: "passthrough" leaves numeric values exactly as-is.
preprocessor = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
    ("num", "passthrough", numerical_features)
    ])
```

## Define Models

```python
# Four candidate classifiers — each will be wrapped in a Pipeline so preprocessing
# is identical and the comparison is truly apples-to-apples
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),         # Linear baseline
    "Decision Tree":       DecisionTreeClassifier(random_state=42),   # Interpretable tree
    "Random Forest":       RandomForestClassifier(random_state=42),   # Ensemble of trees
    "Gradient Boosting":   GradientBoostingClassifier()               # Sequential boosted trees
}

results    = []     # Collects (model_name, accuracy) pairs for the comparison table
best_model = None   # Tracks the best-performing pipeline across all candidates
best_score = 0      # Tracks the highest test accuracy seen so far
```

## Train and Evaluate Models

```python
for name, model in models.items():

    # Build a fresh Pipeline for each candidate: preprocessing → classifier.
    # Fitting inside the Pipeline ensures the encoder is trained on training data only.
    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    # Train on the chronologically earliest 80% of visits
    pipeline.fit(X_train, y_train)

    # Predict on the held-out most-recent 20%
    preds = pipeline.predict(X_test)

    # Fraction of correctly classified risk labels
    acc = accuracy_score(y_test, preds)

    print("\nModel:", name)
    print("Accuracy:", acc)
    print(confusion_matrix(y_test, preds))
    print(classification_report(y_test, preds))

    results.append((name, acc))

    # Promote this pipeline if it outperforms the current best
    if acc > best_score:
        best_score = acc
        best_model = pipeline
```

### Business Context — Multi-Model Comparison
Training four candidate algorithms across the same Pipeline ensures an apples-to-apples comparison (identical preprocessing, identical train/test split). A **time-based split** is used instead of random shuffle because hospital visits are time-ordered: using future data to predict the past would produce artificially inflated scores that don't hold in production.

## Model Comparison Table

```python
# Build a ranked summary table to guide which model to tune further
comparison = pd.DataFrame(results, columns=["Model", "Accuracy"])
comparison.sort_values("Accuracy", ascending=False)
```

### Business Context — Model Selection
Random Forest and Gradient Boosting typically outperform Logistic Regression on healthcare tabular data because:
- Clinical risk is **non-linear** (a 70-year-old ICU patient with chronic condition isn't just the sum of those factors)
- Tree ensembles naturally handle feature interactions without manual polynomial feature creation
- They are robust to skewed distributions and outliers common in billing data

The comparison table guides which model to tune further in the next step.

## Hyperparameter Tuning (Random Forest Example)

```python
# Hyperparameter search space for Random Forest
param_grid = {
    "model__n_estimators":      [100, 200, 500],  # More trees → lower variance, higher cost
    "model__max_depth":         [10, 20, None],   # None = fully grown; risks overfitting
    "model__min_samples_split": [2, 5],           # Lower = finer splits; higher = smoother boundaries
    "model__min_samples_leaf":  [1, 2]            # Min leaf size; guards against memorising noise
}

# Dedicated RF pipeline — reuses the shared preprocessor from above
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(random_state=42))
])

# 3-fold CV with macro F1: treats each risk class equally, penalising poor recall on any class.
# n_jobs=-1 parallelises grid evaluation across all available CPU cores.
grid = GridSearchCV(rf_pipeline, param_grid, cv=3, scoring="f1_macro", n_jobs=-1)

grid.fit(X_train, y_train)

print("Best Parameters:", grid.best_params_)

# Replace best_model with the GridSearchCV winner for final evaluation
best_model = grid.best_estimator_
```

### Business Context — Hyperparameter Tuning
GridSearchCV with 3-fold cross-validation systematically explores the trade-off between:
- **`n_estimators`** — more trees → lower variance but higher training cost
- **`max_depth`** — deeper trees capture complex interactions but risk overfitting
- **`min_samples_split/leaf`** — prevents the model from memorising rare patient combinations

Using `f1_macro` scoring is appropriate here because risk score classes (Low/Medium/High) may be imbalanced — macro F1 penalises poor recall on any class equally, ensuring the model doesn't ignore the clinically critical **High risk** group.

## Final Evaluation

```python
preds = best_model.predict(X_test)

print("Best ModeL:", best_model)
print("Final Accuracy:", accuracy_score(y_test, preds))
print(confusion_matrix(y_test, preds))
print(classification_report(y_test, preds))   # Key metric: High risk Recall
```

### Business Context — Production Readiness Check
The confusion matrix and classification report reveal two key production metrics:
- **High Risk Recall** — proportion of genuinely high-risk visits correctly flagged. Missing a High-risk visit (false negative) has patient safety and liability implications.
- **Macro F1** — balanced performance across all three risk classes; confirms the model serves General, ICU, and ER departments equally.

A model with high overall accuracy but poor High-risk recall is **not fit for clinical deployment**.

## Save Model Artifact

```python
# Persist the full Pipeline (preprocessor + tuned RF) as a single artifact.
# Saving the pipeline — not just the classifier — means main.py can call
# model.predict() on raw string inputs without any manual preprocessing.
joblib.dump(best_model, r".\data\risk_model.pkl", compress=3)
print("risk_model.pkl saved")
```

### Business Context — Model Artifact
`risk_model.pkl` is a **full sklearn Pipeline** containing:
1. ColumnTransformer (OneHotEncoder for categoricals, passthrough for numerics)
2. Tuned RandomForestClassifier

Saving the entire pipeline — not just the classifier — means the API (main.py) can pass raw string inputs directly without any manual preprocessing. This eliminates a common source of train/serve skew in production.

```python
import json

# Define the feature schema as a versioned contract between training and the API.
# The API reads this JSON at startup via utils.load_features() to:
#   1. Know which columns to expect and in which order
#   2. Determine which columns need encoding (dtype "category")
#   3. Supply allowed values for Pydantic input validation
feature_schema = {
    "age":                     {"dtype": "numeric"},
    "gender":                  {"dtype": "category", "values": ["F", "M"]},
    "department":              {"dtype": "category", "values": ["Cardiology", "ER", "General", "ICU", "Neurology", "Orthopedics"]},
    "visit_type":              {"dtype": "category", "values": ["ER", "ICU", "OPD"]},
    "chronic_flag":            {"dtype": "numeric"},
    "length_of_stay_hours":    {"dtype": "numeric"},
    "visit_frequency":         {"dtype": "numeric"},
    "avg_los_per_patient":     {"dtype": "numeric"},
    "days_since_registration": {"dtype": "numeric"}
}

# Wrap in a top-level dict with model metadata
schema = {
    "model_name": "hospital_prediction_model",
    "features": feature_schema
}

# Write to disk — consumed by utils.py at inference time
with open(r".\data\risk_features.json", "w") as f:
    json.dump(schema, f, indent=4)

print("risk_features.json created successfully")
```
